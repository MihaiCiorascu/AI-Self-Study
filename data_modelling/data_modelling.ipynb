### 1.5 Speed of NumPy library

```python
# Sum of elems with FOR vs np.sum()
def sum_of_elems_no_np(x):
  sum = 0
  for nr in x:
    sum += nr
  return sum

def sum_of_elems_np(x):
  return np.sum(x)
```

```python
x = np.random.rand(10000000)
```

```python
%%timeit
sum_of_elems_no_np(x)
```

```python
%%timeit
sum_of_elems_np(x)
```

```python
def matrix_multiplication_no_np(A, B):
  if A.shape[1] != B.shape[0]:
    return False

  N, M, K = A.shape[0], B.shape[1], A.shape[1]

  C = [[0 for i in range(N)] for j in range(M)]
  for i in range(N):
    for j in range(M):
      for k in range(K):
        C[i][j] += A[i][k] * B[k][j]

  return C

def matrix_multiplication_np(A, B):
  if A.shape[1] != B.shape[0]:
    return False

  return A.dot(B) # or A @ B
```

```python
A = np.random.rand(10000).reshape(100, 100)
B = np.random.rand(10000).reshape(100, 100)
```

```python
%%timeit
matrix_multiplication_no_np(A, B)
```

```python
%%timeit
matrix_multiplication_np(A, B)
```

```python
# Usually we will need to cerate new cols to help us in analysing data
# Let s say that we got hired by a scary movies studio to make a market analysis
# In this way, would be useful to see what movies are of type Horror and if they
# were profitable or not

# Firstly, we create a new 'feature' is_horror, that will highlight if the movie
# is horror or not
data['is_horror'] = data['genres'].str.contains('Horror', case=False)
data.head()
```

### 2.3 Data Filtering

```python
# To Locate specific cells, can use 2 methods: .loc and .iloc

# 1 .loc
# data.loc[rows, columns] -> selects from data all the rows from 'rows' and
# keeps only the columns from 'columns'
# KEEP IN MIND that .loc is used for LABELS and also for INDEXES, similar with Lists
# For example, let s say we want to find the first 100 rows and keep only 'MPA' and
# 'Rating' columns, so:
data.loc[:99, ['MPA', 'Rating']] # ATTENTION!!! .loc includes the upper limits range
```

```python
# 2 .iloc
# data.iloc[rows, columns] -> selects from data all the rows from 'rows' and
# keeps only the columns from 'columns'
# Compared to .loc, .iloc uses ONLY INDEXES
data.iloc[:100, [5, 6]] # the selection from above, but with .iloc
```

```python
# Coming back to the horror movie example
# To filter the movies, there are 2 methods: .loc and .apply()

# .loc is maybe the easiser one, but less flexible
good_movies = data.loc[data['is_horror'] & data['is_profitable']] # OBSERVATION
                                                        # .loc can be forgiven
good_movies.head()
```

```python
# .apply() is much flexible, because uses a filter function, but takes more code
# to write

def is_good(movie):
  return movie['is_horror'] and movie['is_profitable']

good_movies = data[data.apply(is_good, axis=1)]
good_movies.head()
```

# 3. Matplotlib.pyplot

```python
import matplotlib.pyplot as plt
```

### 3.1 Liniar Charts

The Liniar Chart is used to display a continuous relationship between 2 vars. Usually, shows how a var (y-axis) modifies another (x-axis)

```python
x = np.linspace(0, 10, 100) # 100 values in [0, 10]
y = np.sin(x)
plt.figure(figsize=(6, 4))

plt.plot(x, y, label="sin(x)", color="blue")
plt.title("Liniar Graph")

plt.xlabel("x")
plt.ylabel("y")

plt.legend()
plt.grid(True)
plt.show()
```

### 3.2 Scatter Chart

The Scatter Chart uses dots to represent dataset values. Each dot is an observation with coordinates defined by 2 vars

```python
x = np.random.rand(50)
y = np.random.rand(50)
plt.figure(figsize=(6, 4))

plt.scatter(x, y, color="red", alpha=0.7)
plt.title("Scatter Graph")

plt.xlabel("x")
plt.ylabel("y")

plt.show()
```

### 3.3 Bar Chart

The Bar Chart is for categorical data using rectangular bars, where every bar's length is proportional to its value

```python
categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]

plt.figure(figsize=(6, 4))
plt.bar(categories, values, color='green')

plt.title("Bar Chart")
plt.xlabel("Categories")
plt.ylabel("Values")

plt.show()
```

### 3.4 Histogram

A Histogram is used for the frequency distribution of a value. Data is grouped by intervals (bins)

```python
distribution = np.random.randn(1000) # 1000 values from a normal distribution

plt.figure(figsize=(6, 4))
plt.hist(distribution, bins=20, color='purple', alpha=0.8)

plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```
